```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import datetime, timedelta


```

```python
def get_all_stock_data():
    YEARS = 10
    MIN_PRICE = 8
    MAX_PRICE = 200

    # Load and format tickers
    tickers = (
        pd.read_csv("data/tsx_tickers_extracted.csv")["Ticker"]
        .dropna().unique()
    )
    tickers = [f"{t.strip().upper()}.TO" for t in tickers]

    # Date range
    end_date = datetime.today()
    start_date = end_date - timedelta(days=YEARS * 365)

    final_df = pd.DataFrame()

    for ticker in tickers:
        try:
            data = yf.download(ticker, start=start_date, end=end_date, progress=False, auto_adjust=True)

            if data.empty:
                continue

            # Flatten MultiIndex columns (yfinance >= 0.2.x)
            if isinstance(data.columns, pd.MultiIndex):
                data.columns = data.columns.get_level_values(0)

            # Get Close series (auto_adjust=True makes Close == Adj Close)
            series = data["Close"]

            # Handle duplicate columns after flatten
            if isinstance(series, pd.DataFrame):
                series = series.iloc[:, 0]

            series = series.dropna()

            # Filter by latest price
            latest_price = float(series.iloc[-1])
            if not (MIN_PRICE <= latest_price <= MAX_PRICE):
                continue

            final_df[ticker] = series

        except Exception as e:
            c=1
            # print(f"Error with {ticker}: {e}")

    # Sort descending and save
    final_df.index.name = "Date"
    final_df.sort_index(ascending=False).to_csv("data/all_stock_data.csv")
```
